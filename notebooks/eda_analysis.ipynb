```python
import pandas as pd

df = pd.read_csv("csfloat_butterfly_knife.csv")
df = df[df["float_value"] > 0].copy()

# Podejrzane oferty - powyżej 99 percentyla
percentyl_99 = df["price_usd"].quantile(0.99)
print(f"99. percentyl ceny: ${percentyl_99:,.2f}")
print(f"\nObserwacje powyżej 99. percentyla: {len(df[df['price_usd'] > percentyl_99])}")
print("\nNajdroższe 20 ofert:")
print(
    df.nlargest(20, "price_usd")[
        ["market_hash_name", "float_value", "price_usd", "is_stattrak"]
    ].to_string()
)
```

```python
# Znane Blue Gem pattern seedy dla Butterfly Knife Case Hardened
blue_gem_seeds = [387, 601, 670, 363, 268]

# Maska chroniąca Blue Gemy
is_blue_gem = (
    df["market_hash_name"].str.contains("Case Hardened") & 
    df["paint_seed"].isin(blue_gem_seeds)
)

# Fałszywe oferty - wysoka cena + normalny float + nie Blue Gem
fake_mask = (
    (df["price_usd"] > 40000) & 
    (df["float_value"] > 0.005) & 
    (~is_blue_gem)
)

print("Obserwacje do usunięcia:")
print(df[fake_mask][["market_hash_name", "paint_seed", "float_value", "price_usd"]].to_string())
print(f"\nLiczba usuniętych: {fake_mask.sum()}")

print("\nChronione Blue Gemy:")
print(df[is_blue_gem][["market_hash_name", "paint_seed", "float_value", "price_usd"]].to_string())
```
